# 🌱 My First Strands Agent with Tools

A beginner-friendly example of building a **Strands Agents** agent in an AWS SageMaker AI Jupyter notebook.

### What this notebook does
We will create an agent that can:
1. Use a `calculate` tool for simple arithmetic.
2. Use a `get_weather` tool with sample weather data.
3. Decide which tool to use based on the user's question.

> **Note:** The weather tool uses sample data so that the notebook stays simple and does not require a weather API.

## Step 2 — Import the required libraries

`Agent` creates our AI agent and `tool` turns a normal Python function into a tool that the agent can call.

```python
from strands import Agent, tool

print("Strands Agents imported successfully!")
```

## Step 3 — Create a calculator tool

The agent can call this Python function whenever it needs to perform a calculation.

```python
@tool
def calculate(expression: str) -> str:
    """Calculate a simple arithmetic expression.

    Args:
        expression: A mathematical expression such as "25 * 4" or "100 / 5".
    """
    try:
        # Keep this example limited to basic arithmetic.
        allowed = set("0123456789+-*/(). %")
        if not all(char in allowed for char in expression):
            return "I can only calculate basic arithmetic expressions."

        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except Exception as e:
        return f"Could not calculate the expression: {e}"
```

## Step 4 — Create a simple weather tool

This is deliberately a **mock tool**. Later, you can replace it with a real weather API.

```python
@tool
def get_weather(city: str) -> str:
    """Return sample weather information for a city.

    Args:
        city: Name of the city.
    """
    weather_data = {
        "bengaluru": "Bengaluru: 27°C, partly cloudy",
        "mumbai": "Mumbai: 30°C, humid",
        "delhi": "Delhi: 32°C, sunny",
        "chennai": "Chennai: 31°C, partly cloudy",
    }

    return weather_data.get(
        city.strip().lower(),
        f"No sample weather data is available for {city}."
    )
```

## Step 5 — Create the Strands Agent

The agent receives the two tools. It can decide when a tool is useful.

```python
agent = Agent(
    tools=[calculate, get_weather],
    system_prompt="""You are a friendly beginner assistant.
Use the available tools when they are useful.
Explain your answer clearly and briefly."""
)

print("Agent created successfully!")
```

## Step 6 — Ask the agent a calculation question

```python
response = agent("What is 25 * 8?")
print(response)
```

## Step 7 — Ask the agent about weather

```python
response = agent("What is the weather like in Bengaluru?")
print(response)
```

## Step 8 — Ask a question that uses both tools

This demonstrates a more agentic workflow.

```python
response = agent(
    "What is the sample temperature in Bengaluru, and what is that temperature "
    "in Fahrenheit? Show the calculation."
)
print(response)
```

## Step 9 — Try your own question

Change the text and run the cell again.

```python
question = "Calculate 150 / 6 and explain the answer in one sentence."
response = agent(question)
print(response)
```
